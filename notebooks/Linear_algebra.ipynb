```python
import numpy as np
np.__version__
```

```python
x = np.array([1, 2, 3], dtype=np.float32)
y = np.array([[1, 2, 3]], dtype=np.float32)
z = np.array([[1], [2], [3]], dtype=np.float32)

print(x)
print(y)
```

```python
x = np.array([1, 2, 3], dtype=np.float32)
y = np.asarray([1., 2., 3.])

print(f"Тип данных в массиве x -> {x.dtype}")
print(f"Тип данных в массиве y -> {y.dtype}")
```

```python
x = np.array([1, 2, 3], dtype=np.float32)
x_T = x.T

print(x.shape)
```

```python
result = x @ x

print(result)
print(result.shape)
```

```python
result = y @ y.T

print(result)
print(result.shape)
print(type(result))
```
