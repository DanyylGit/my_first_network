# Математический анализ для нейронных сетей: Производные и градиенты

## 1. Определение и геометрический смысл производной

* **Предельное определение производной:**
$$f'(x) = \frac{df}{dx} = \lim_{\Delta x \to 0} \frac{f(x + \Delta x) - f(x)}{\Delta x}$$

* **Геометрический смысл:**
$$f'(x_0) = \tan(\alpha)$$
где $\alpha$ — угол наклона касательной к графику функции $f(x)$ в точке $x_0$.

---

## 2. Табличные производные основных функций

$$
\begin{aligned}
\frac{d}{dx}(c) &= 0 \quad (c = \text{const}) \\
\frac{d}{dx}(x^n) &= n x^{n-1} \\
\frac{d}{dx}(e^x) &= e^x \\
\frac{d}{dx}(\ln x) &= \frac{1}{x} \\
\frac{d}{dx}(\sin x) &= \cos x \\
\frac{d}{dx}(\cos x) &= -\sin x \\
\frac{d}{dx}(\sigma(x)) &= \sigma(x) \cdot (1 - \sigma(x)) \quad \text{где } \sigma(x) = \frac{1}{1 + e^{-x}}
\end{aligned}
$$

---

## 3. Основные правила дифференцирования

1. **Вынос константы:**
$$(c \cdot f(x))' = c \cdot f'(x)$$

2. **Производная суммы и разности:**
$$(f(x) \pm g(x))' = f'(x) \pm g'(x)$$

3. **Производная произведения:**
$$(f(x) \cdot g(x))' = f'(x) g(x) + f(x) g'(x)$$

4. **Производная частного:**
$$\left( \frac{f(x)}{g(x)} \right)' = \frac{f'(x) g(x) - f(x) g'(x)}{g(x)^2}$$

5. **Производная сложной функции (Chain Rule):**
$$\frac{d}{dx} f(g(x)) = \frac{\partial f}{\partial g} \cdot \frac{\partial g}{\partial x}$$

---

## 4. Частные производные и векторный анализ

* **Частная производная** функции многих переменных $L(w_1, w_2, \dots, w_n)$ (все остальные переменные считаются константами):
$$\frac{\partial L}{\partial w_i}$$

* **Градиент** (вектор частных производных, указывает направление наискорейшего роста):
$$\nabla L(\mathbf{w}) = \left( \frac{\partial L}{\partial w_1}, \frac{\partial L}{\partial w_2}, \dots, \frac{\partial L}{\partial w_n} \right)^T$$

* **Антиградиент** (направление наискорейшего спуска):
$$-\nabla L(\mathbf{w})$$

---

## 5. Применение в оптимизации нейросетей

* **Формула обновления параметров (Градиентный спуск):**
$$\mathbf{w}_{\text{new}} = \mathbf{w}_{\text{old}} - \lambda \cdot \nabla L(\mathbf{w}_{\text{old}})$$
где $\lambda > 0$ — скорость обучения (learning rate).

* **Обратный проход в вычислительном графе (Backpropagation):**
$$\frac{\partial L}{\partial x} = \frac{\partial L}{\partial f} \cdot \frac{\partial f}{\partial x}$$
где $\frac{\partial L}{\partial f}$ — градиент, пришедший справа по графу.

# Линейная алгебра: Операции над векторами

## 1. Транспонирование вектора

* **Определение:** Транспонирование — это операция поворота вектора (или матрицы), при которой его строки становятся столбцами, а столбцы — строками.
* **Обозначение:** Верхний индекс $T$ ($\mathbf{x}^T$).

### Формула и пример:
Если дан вектор-столбец размерности $(n \times 1)$:
$$\mathbf{x} = \begin{pmatrix} x_1 \\ x_2 \\ x_3 \end{pmatrix}$$

То результат его транспонирования — вектор-строка размерности $(1 \times n)$:
$$\mathbf{x}^T = \begin{pmatrix} x_1 & x_2 & x_3 \end{pmatrix}$$

#### Числовой пример:
$$\mathbf{v} = \begin{pmatrix} 4 \\ -2 \\ 7 \end{pmatrix} \quad \xrightarrow{\text{транспонирование}} \quad \mathbf{v}^T = \begin{pmatrix} 4 & -2 & 7 \end{pmatrix}$$

> ⚠️ **Важно для NumPy:** 
> В NumPy одномерный массив `x = np.array([1, 2, 3])` имеет форму `(3,)` (у него нет ориентации строки или столбца). Операция `x.T` **не меняет** такой массив. 
> Чтобы транспонирование сработало, массив должен быть двумерным: `x.reshape(-1, 1)` имеет форму `(3, 1)`.

---

## 2. Скалярное произведение векторов (Dot Product)

* **Определение:** Операция над двумя векторами **одинаковой размерности**, при которой перемножаются их соответствующие элементы, а полученные произведения складываются.
* **Результат:** Всегда является **одним числом (скаляром)**.

### Формулы:
1. **Через сумму попарных произведений:**
   $$\mathbf{a} \cdot \mathbf{b} = \sum_{i=1}^{n} a_i b_i = a_1 b_1 + a_2 b_2 + \dots + a_n b_n$$

2. **В матричном виде («строка на столбец»):**
   $$\mathbf{a}^T \mathbf{b} = \begin{pmatrix} a_1 & a_2 & \dots & a_n \end{pmatrix} \begin{pmatrix} b_1 \\ b_2 \\ \vdots \\ b_n \end{pmatrix} = a_1 b_1 + a_2 b_2 + \dots + a_n b_n$$

#### Числовой пример:
Пусть даны весовой вектор $\mathbf{w}$ и вектор признаков $\mathbf{x}$:
$$\mathbf{w} = \begin{pmatrix} 2 \\ -1 \\ 4 \end{pmatrix}, \quad \mathbf{x} = \begin{pmatrix} 3 \\ 5 \\ 2 \end{pmatrix}$$

Вычисляем скалярное произведение:
$$\mathbf{w} \cdot \mathbf{x} = (2 \cdot 3) + ((-1) \cdot 5) + (4 \cdot 2) = 6 - 5 + 8 = \mathbf{9}$$

---

## 3. Применение в нейронных сетях

Взвешенная сумма входов искусственного нейрона до применения функции активации рассчитывается именно через **скалярное произведение** вектора весов $\mathbf{w}$ на вектор входных сигналов $\mathbf{x}$ с добавлением смещения $b$:

$$z = \mathbf{w}^T \mathbf{x} + b = w_1 x_1 + w_2 x_2 + \dots + w_n x_n + b$$

---

## 4. Сравнение: Скалярное vs Поэлементное умножение

* **Скалярное умножение ($\mathbf{a} \cdot \mathbf{b}$ или $\mathbf{a}^T \mathbf{b}$):** перемножает элементы и **складывает** их. Возвращает **число**.
* **Поэлементное умножение ($\mathbf{a} \odot \mathbf{b}$ / Hadamard Product):** только перемножает парные элементы. Возвращает **вектор той же размерности**:
  $$\begin{pmatrix} a_1 \\ a_2 \end{pmatrix} \odot \begin{pmatrix} b_1 \\ b_2 \end{pmatrix} = \begin{pmatrix} a_1 \cdot b_1 \\ a_2 \cdot b_2 \end{pmatrix}$$
